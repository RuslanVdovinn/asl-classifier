```python
import torch
import torchvision.models as models
```

```python
model = models.resnet18(pretrained=True)
print(model)

print(f"\nПараметров: {sum(p.numel() for p in model.parameters()):,}")

model.fc = torch.nn.Linear(model.fc.in_features, 29)
print(f"\nПосле замены головы:")
print(f"Параметров: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# ResNet18
resnet = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
print(f"ResNet18: {sum(p.numel() for p in resnet.parameters()):,}")

# EfficientNet-B0
effnet = models.efficientnet_b0(pretrained=True)
print(f"EfficientNet-B0: {sum(p.numel() for p in effnet.parameters()):,}")
```
