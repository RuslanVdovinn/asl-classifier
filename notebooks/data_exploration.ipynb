```python
from pathlib import Path

DATA_DIR = Path("data/asl_alphabet_train")
classes = sorted([d.name for d in DATA_DIR.iterdir() if d.is_dir()])
print(f"Классов: {len(classes)}")
print(f"Классы: {classes}")

total = sum(len(list((DATA_DIR / cls).glob("*.jpg"))) for cls in classes)
print(f"Всего изображений: {total}")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms

img = Image.open("../data/asl_alphabet_train/A/A1.jpg").convert("RGB")

augment = transforms.Compose([
    transforms.Resize((200, 200)),
    transforms.RandomRotation(15),
    transforms.RandomAffine(0, translate=(0.1, 0.1)),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
])

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax in axes.flat:
    aug_img = augment(img)
    ax.imshow(aug_img)
    ax.axis("off")
plt.suptitle("Аугментации одного изображения (буква A)", fontsize=16)
plt.tight_layout()
plt.show()
```
